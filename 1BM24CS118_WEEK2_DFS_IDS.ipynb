```python
import sys

# Remove python stack limits for deep search trees
sys.setrecursionlimit(200000)

class PuzzleNode:
    def __init__(self, state, parent=None, move=None):
        self.state = state      # Tuple of 9 integers
        self.parent = parent    # Link to the parent node
        self.move = move        # The move that led to this state

    def get_blank_index(self):
        return self.state.index(0)

    def generate_children(self):
        children = []
        blank = self.get_blank_index()
        row, col = blank // 3, blank % 3

        # Define valid moves: (row_change, col_change, move_name)
        moves = [(-1, 0, 'Up'), (1, 0, 'Down'), (0, -1, 'Left'), (0, 1, 'Right')]

        for dr, dc, move_name in moves:
            new_row, new_col = row + dr, col + dc
            if 0 <= new_row < 3 and 0 <= new_col < 3:
                new_blank = new_row * 3 + new_col
                # Swap blank with the adjacent number
                new_state = list(self.state)
                new_state[blank], new_state[new_blank] = new_state[new_blank], new_state[blank]
                children.append(PuzzleNode(tuple(new_state), self, move_name))
        return children

def dfs_solve(start_state, goal_state):
    """
    Solves the 8-puzzle using an unrestricted Depth-First Search.
    """
    stack = [PuzzleNode(start_state)]
    visited = {start_state: 0} # Maps state to the depth it was found

    while stack:
        current_node = stack.pop()

        if current_node.state == goal_state:
            return current_node

        # Calculate current depth by tracing parents
        depth = 0
        p = current_node.parent
        while p:
            depth += 1
            p = p.parent

        for child in current_node.generate_children():
            # If not visited, or found at a shallower depth, explore it
            if child.state not in visited or depth + 1 < visited[child.state]:
                visited[child.state] = depth + 1
                stack.append(child)

    return None

def print_path(node):
    path = []
    current = node
    while current:
        path.append(current)
        current = current.parent
    path.reverse()

    print(f"\nSolved in {len(path) - 1} moves:")
    for step, n in enumerate(path):
        if n.move:
            print(f"Move {step}: {n.move}")
        else:
            print("Initial State:")

        # Print 3x3 grid
        s = n.state
        print(f"{s[0]} {s[1]} {s[2]}\n{s[3]} {s[4]} {s[5]}\n{s[6]} {s[7]} {s[8]}\n")

def get_user_input(prompt):
    print(prompt)
    print("Enter 9 numbers (0-8) separated by spaces. Example: 1 2 3 4 0 5 6 7 8")
    while True:
        try:
            user_in = input("> ").strip().split()
            if len(user_in) != 9:
                raise ValueError("You must enter exactly 9 numbers.")
            state = tuple(int(x) for x in user_in)
            if sorted(list(state)) != list(range(9)):
                raise ValueError("Numbers must be unique and range from 0 to 8.")
            return state
        except ValueError as e:
            print(f"Invalid input: {e}. Please try again.")

if __name__ == "__main__":
    print("--- 8-PUZZLE DFS SOLVER ---")
    start = get_user_input("Step 1: Enter the INITIAL state")
    goal = get_user_input("\nStep 2: Enter the GOAL state")

    print("\nSearching for a solution using DFS...")
    solution = dfs_solve(start, goal)

    if solution:
        print_path(solution)
    else:
        print("\nNo solution found.")
```

```python
def get_moves(state):
    """Generates all valid next states from the current board state."""
    zero_idx = state.index(0)
    row, col = divmod(zero_idx, 3)

    moves = []
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)] # Up, Down, Left, Right

    for dr, dc in directions:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_idx = new_row * 3 + new_col
            state_list = list(state)
            state_list[zero_idx], state_list[new_idx] = state_list[new_idx], state_list[zero_idx]
            moves.append(tuple(state_list))

    return moves

def depth_limited_dfs(current, goal, limit, path, visited):
    """A recursive, depth-limited DFS helper function."""
    if current == goal:
        return path

    if limit <= 0:
        return None

    visited.add(current)

    for next_state in get_moves(current):
        if next_state not in visited:
            result = depth_limited_dfs(next_state, goal, limit - 1, path + [next_state], visited)
            if result is not None:
                return result

    visited.remove(current)
    return None

def iddfs(start, goal):
    """Main IDDFS loop that increments the depth limit progressively."""
    depth = 0
    while True:
        print(f"Searching at depth limit: {depth}...")
        visited = set()
        result = depth_limited_dfs(start, goal, depth, [start], visited)

        if result is not None:
            return result, depth

        depth += 1

def print_board(state):
    """Prints a 1D tuple formatted beautifully as a 3x3 board."""
    for i in range(0, 9, 3):
        print(f"  {state[i]} {state[i+1]} {state[i+2]}")
    print()

def get_user_input(prompt_name):
    """Prompts the user for a valid 8-puzzle board configuration."""
    while True:
        print(f"\nEnter the {prompt_name} state.")
        print("Provide 9 numbers (0-8) separated by spaces row by row. Use '0' for the blank space.")
        print("Example: 1 2 3 4 0 6 7 5 8")

        try:
            user_str = input(">> ").strip()
            # Convert input text into a list of integers
            digits = [int(x) for x in user_str.split()]

            # Validation checks
            if len(digits) != 9:
                print("❌ Error: You must enter exactly 9 numbers.")
                continue
            if set(digits) != set(range(9)):
                print("❌ Error: The numbers must be exactly from 0 to 8 with no duplicates.")
                continue

            return tuple(digits)
        except ValueError:
            print("❌ Error: Invalid characters. Please enter numbers only.")

# --- Main Driver Script ---
if __name__ == "__main__":
    print("=== 8-Puzzle Solver using Iterative Deepening DFS (IDDFS) ===")

    start_state = get_user_input("STARTING")
    goal_state = get_user_input("GOAL")

    print("\n--- Initializing Search ---")
    print("Initial State:")
    print_board(start_state)
    print("Goal State:")
    print_board(goal_state)

    solution_path, final_depth = iddfs(start_state, goal_state)

    print(f"\n✨ Success! Optimal solution found at depth: {final_depth}")
    print(f"Total moves required: {len(solution_path) - 1}")
    print("\n--- Step-by-Step Path ---")

    for step, state in enumerate(solution_path):
        if step == 0:
            print(f"Step {step} (Start):")
        else:
            print(f"Step {step}:")
        print_board(state)

```
