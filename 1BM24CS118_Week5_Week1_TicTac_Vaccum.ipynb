1. Implementing Tic-Tac-Toe

```python
board = ["1", "2", "3",
         "4", "5", "6",
         "7", "8", "9"]

players = {
    "X": "Player 1",
    "O": "Player 2"
}

winning_combinations = (
    (0, 1, 2),
    (3, 4, 5),
    (6, 7, 8),
    (0, 3, 6),
    (1, 4, 7),
    (2, 5, 8),
    (0, 4, 8),
    (2, 4, 6)
)

occupied = set()

def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()

def check_winner(symbol):
    for combination in winning_combinations:
        if all(board[position] == symbol for position in combination):
            return True
    return False

def make_move(symbol):
    while True:
        choice = input(
            f"{players[symbol]} ({symbol}), choose a position (1-9): "
        )

        if not choice.isdigit():
            print("Please enter a valid number.")
            continue

        position = int(choice) - 1

        if position < 0 or position > 8:
            print("Please choose a position between 1 and 9.")
            continue

        if position in occupied:
            print("This position is already occupied.")
            continue

        board[position] = symbol
        occupied.add(position)
        break

print("Welcome to Tic-Tac-Toe!")

current_player = "X"

while True:
    display_board()

    make_move(current_player)

    if check_winner(current_player):
        display_board()
        print(players[current_player], "wins!")
        break

    if len(occupied) == 9:
        display_board()
        print("It's a draw!")
        break

    current_player = "O" if current_player == "X" else "X"

print("\nThanks for playing!")
```

2. Implementing Vaccum cleaner (Simple Reflex agent)

```python
class VacuumEnvironment:
    def __init__(self, status_a='Dirty', status_b='Dirty'):
        self.locations = {'A': status_a, 'B': status_b}
        self.cost = 0
    def get_status(self, location):
        return self.locations[location]

    def update_environment(self, location, action):
        """Applies the action to the environment and updates the path cost."""
        if action == "Suck":
            self.locations[location] = "Clean"
            self.cost += 1
            print(f"Action: [Suck] in Room {location}. (Cost +1)")
        elif action == "Move_Right":
            self.cost += 1
            print(f"Action: [Move_Right] to Room B. (Cost +1)")
        elif action == "Move_Left":
            self.cost += 1
            print(f"Action: [Move_Left] to Room A. (Cost +1)")

    def are_all_rooms_clean(self):
        return self.locations['A'] == 'Clean' and self.locations['B'] == 'Clean'


class SimpleReflexAgent:
    def __init__(self, initial_location='A'):
        self.current_location = initial_location

    def interpret_input(self, location_status):
        """Condition-Action Rules (Simple Reflex Logic)"""
        if location_status == 'Dirty':
            return 'Suck'
        elif self.current_location == 'A':
            return 'Move_Right'
        elif self.current_location == 'B':
            return 'Move_Left'

    def run_step(self, env):
        """Perceive, decide, act, and transition to the next state."""
        current_status = env.get_status(self.current_location)
        print(f"\nPercept -> Location: {self.current_location}, Status: {current_status}")

        action = self.interpret_input(current_status)

        env.update_environment(self.current_location, action)

        if action == 'Move_Right':
            self.current_location = 'B'
        elif action == 'Move_Left':
            self.current_location = 'A'


def get_user_choice(room_name):
    """Helper function to ensure valid user input ('Dirty' or 'Clean')."""
    while True:
        choice = input(f"Enter status for Room {room_name} (1 for Dirty, 2 for Clean): ").strip()
        if choice == '1':
            return 'Dirty'
        elif choice == '2':
            return 'Clean'
        print("Invalid input! Please enter 1 or 2.")


if __name__ == "__main__":
    print("--- Configure Vacuum Environment ---")
    status_a = get_user_choice('A')
    status_b = get_user_choice('B')

    environment = VacuumEnvironment(status_a=status_a, status_b=status_b)
    agent = SimpleReflexAgent(initial_location='A')

    print("\n--- Starting Vacuum Cleaner Simulation ---")
    print(f"Initial State: {environment.locations}")

    step = 1
    while not environment.are_all_rooms_clean() and step <= 10:
        print(f"\n--- Step {step} ---")
        agent.run_step(environment)
        step += 1

    print("\n--- Simulation Finished ---")
    print(f"Final Environment State: {environment.locations}")
    print(f"Total Path Cost: {environment.cost}")
```
